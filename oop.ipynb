```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"Person(name={self.name}, age={self.age})"

bob = Person("Bob", 13)
print("Name:", bob.name)
print("age:", bob.age)
bob.company = "Samsung"
print("Company:", bob.company)
print(bob)
```

```python
class Hero:
   #конструктор класса
   def __init__(self, name, health, armor):
       self.name = name #строка
       self.health = health #число
       self.armor = armor #число
   #печать параметров персонажа:
   def print_info(self):
       print('Уровень здоровья:', self.health)
       print('Класс брони:', self.armor, '\n')

class Warrior(Hero):
   def hello(self):
       print('-> НОВЫЙ ГЕРОЙ. Верхом на коне появился бравый воин по имени', self.name)
       self.print_info()

   def attack(self, enemy):
       print('-> УДАР! Храбрый воин', self.name, 'атакует', enemy.name, 'мечом!')
       enemy.armor -= 15 #сила удара для класса Воин
       if enemy.armor < 0:
           enemy.health += enemy.armor
           enemy.armor = 0
       print('Страшный удар обрушился на противника. \nТеперь его броня: ' +
           str(enemy.armor) + ', а уровень здоровья: ' + str(enemy.health) + '\n')

class Magician(Hero):
   def hello(self):
       print('-> НОВЫЙ ГЕРОЙ. Появился маг по имени', self.name)
       self.print_info()

   def cast_spell(self, enemy):
       print('-> УДАР! Маг', self.name, 'накладывает заклинание на', enemy.name)
       enemy.armor -= 15
       if enemy.armor < 0:
           enemy.health += enemy.armor
           enemy.armor = 0
       print('Страшный удар обрушился на противника. \nТеперь его броня: ' +
           str(enemy.armor) + ', а уровень здоровья: ' + str(enemy.health) + '\n')
```

```python
warrior1 = Warrior('Henry', 100, 50)
enemy1 = Hero('Enemy', 100, 50)
warrior1.hello()
warrior1.attack(enemy1)
```

```python
magician1 = Magician('Bob', 100, 20)
magician1.hello()
magician1.cast_spell(enemy1)
```
