```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"Person(name={self.name}, age={self.age})"

bob = Person("Bob", 13)
print("Name:", bob.name)
print("age:", bob.age)
bob.company = "Samsung"
print("Company:", bob.company)
print(bob)
```
