```python
import turtle

class Line:

    def __init__(self, x1, y1, x2, y2, color='black'):
        self._x1 = x1
        self._y1 = y1
        self._x2 = x2
        self._y2 = y2
        self._color = color

    def set_color(self, color):
        self._color = color

    def get_color(self):
        return self._color

    def set_coords(self, x1, y1, x2, y2):
        self._x1 = x1
        self._y1 = y1
        self._x2 = x2
        self._y2 = y2

    def get_coords(self):
        return self._x1, self._y1, self._x2, self._y2

    def draw(self):
        turtle.up()
        turtle.goto(self._x1, self._y1)
        turtle.down()
        turtle.pencolor(self._color)
        turtle.goto(self._x2, self._y2)

# Создание объекта линии

line = Line(0, 0, 50, 50)
# Изменение цвета линии

line.set_color('red')
# Рисование линии

line.draw()
```
