4. Найдите дисперсию и стандартное отклонение массива [2, 4, 6, 8, 10].

```python
arr4 = np.array([2, 4, 6, 8, 10])
print("дисперсия =", arr4.var(), ", стандартное отклонение =", arr4.std())
```

5. Создайте матрицу 3×3, найдите сумму её элементов по диагонали.

```python
m5 = np.arange(1, 10).reshape(3, 3)
print("матрица\n", m5)
print("сумма по главной диагонали =", np.trace(m5))
```

6. Объедините три массива в один одномерный.

```python
b1 = np.array([1, 2, 3])
b2 = np.array([4, 5, 6])
b3 = np.array([7, 8, 9])
merged = np.concatenate([b1, b2, b3])
print(merged)
```

7. Создайте массив 5×5 и вычислите кумулятивную сумму по строкам.

```python
m7 = np.arange(1, 26).reshape(5, 5)
sum_rows = np.cumsum(m7, axis=1)
print("исходная матрица\n", m7)
print("   кумулятивная сумма по строкам\n", sum_rows)
```

8. Из массива [1, 2, 3, 4, 5, 6] получите массив [1, 4, 9, 16, 25, 36] без циклов.

9. Замените все NaN в массиве [1, np.nan, 3, np.nan, 5] на 0.

```python
arr9 = np.array([1, np.nan, 3, np.nan, 5])
cleaned = np.nan_to_num(arr9, nan=0)
print(cleaned)
```

10. Создайте матрицу 4×4 и поверните её на 90 градусов.

```python
m10 = np.arange(1, 17).reshape(4, 4)
rotated = np.rot90(m10)
print("исходная\n", m10)
print("повёрнутая на 90°\n", rotated)
```
