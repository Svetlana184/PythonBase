```python
from random import randint
mas = tuple(randint(1, 100) for i in range (randint(10, 21)))
for i in mas:
    print(i, end=" ")
```

```python
s = 0
for i in mas:
    s += i
print(s)
```

Двумерные массивы

```python
n,m = 5,6
matrix = tuple(tuple(randint(1, 100) for j in range (m)) for i in range (n))
summa_d = 0
for i in matrix:
    print(*i)
for i in range(len(matrix)):
    for j in range(len(matrix)):
        if i ==j:
            summa_d += matrix[i][j]
print("s = ", summa_d)
```
