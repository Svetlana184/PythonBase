На основании официальных статистических данных о курсе акции, которые были использованы при решении задания 3.2 (предварительно необходимо заменить информацию о ценах для нескольких строк на следующие символы «-», « », «н/д»: всего заменить не более 3–4 значений, при этом в одной из строк должно быть не менее 2 таких символов), используя функцию pandas.read_csv, создайте набор данных, в котором решите следующие проблемы:
а) проблему индексирования: определите имена столбцов либо по данным, содержащимся в первой строке исходного файла (аргумент header), либо задайте имена столбцов самостоятельно (аргумент names), а также исключите из результирующего набора данных столбец с данными об изменении цен (аргумент usecols);
б) проблему выведения типа и преобразование данных: определите тип данных для значений, содержащихся в столбцах (для столбца с данными о датах установить тип данных datetime64, для цен и объёмов продаж – float64), при необходимости конвертируя значения (аргументы dtype и converters); определите последовательность значений, интерпретируемых как маркеры отсутствующих данных (аргумент na_values);
в) проблему «грязных данных»: определите символ, который нужно распознать как десятичную точку (аргумент decimal).

```python
CSV_SRC = "MKB.csv"
CSV_DIRTY = "MKB_dirty.csv"
with open(CSV_SRC, "r", encoding="utf-8-sig") as f:
    lines = f.readlines()
def replace_field(line: str, col_idx: int, new_val: str) -> str:
    parts = [p.strip().strip('"') for p in line.rstrip("\n").split(",")]
    parts[col_idx] = new_val
    return ",".join(f'"{p}"' for p in parts) + "\n"

lines[3] = replace_field(lines[3], 3, "н/д")
lines[3] = replace_field(lines[3], 4, "-")

lines[6] = replace_field(lines[6], 2, " ")
lines[10] = replace_field(lines[10], 5, "-")

with open(CSV_DIRTY, "w", encoding="utf-8") as f:
    f.writelines(lines)

print("Создан «грязный» файл:", CSV_DIRTY)
print("Пример «испорченных» строк:")
for i in (3, 6, 10):
    print("  ", lines[i].rstrip())
```

```python
col_names = ["Date", "Close", "Open", "High", "Low", "Volume"]

df = pd.read_csv(
    CSV_DIRTY,
    encoding="utf-8-sig",
    sep=",",
    header=0,
    names=col_names,
    usecols=range(6),
    decimal=",",
    na_values=["-", " ", "н/д", ""],
    dtype={
        "Date": "string",
        "Close": "float64",
        "Open": "float64",
        "High": "float64",
        "Low": "float64",
        "Volume": "string",
    },
    converters={
        "Date": lambda x: pd.to_datetime(x, format="%d.%m.%Y", errors="coerce"),
        "Volume": lambda x: (
            np.nan if pd.isna(x) or str(x).strip() in ("-", "", "н/д") else
            float(str(x).replace(",", ".").replace("M", "e6")
                               .replace("K", "e3"))
        ),
    },
    thousands=None,
)

print("\nТипы данных после read_csv:")
print(df.dtypes)

print("\nПервые 12 строк:")
print(df.head(12))

print("\nПропуски по столбцам:")
print(df.isna().sum())
```

На основании набора данных, полученного в п.1, осуществите:
а) сортировку значений по столбцу с данными об объёмах продаж по убыванию (sort_values());
б) ранжирование значений по столбцу с данными о цене акции на момент открытия (rank());
в) получение сводной статистики по динамике цен на акцию (describe());
г) вычисление корреляции и ковариации по цене закрытия и объёму продаж (corr() и cov());
д) обработку отсутствующих данных:

исключите из набора данных строки, в которых отсутствует 2 и более значений (dropna() с параметром how);

заполните пустые значения в наборе данных средним арифметическим значением (для ценовых данных – по данным строк, по данным об объёмах продаж – по данным столбца) (fillna()).

```python
df_sorted_vol = df.sort_values(by="Volume", ascending=False)
print("\nСортировка по объёму (убывание), топ-5:")
print(df_sorted_vol.head(5)[["Date", "Volume"]])

df["Open_rank"] = df["Open"].rank(method="min", ascending=True)
print("\nРанжирование по Open (первые 10):")
print(df[["Date", "Open", "Open_rank"]].head(10))

print("\ndescribe() по ценовым столбцам:")
print(df[["Open", "High", "Low", "Close", "Volume"]].describe())

pair = df[["Close", "Volume"]].dropna()
corr_cv = pair["Close"].corr(pair["Volume"])
cov_cv  = pair["Close"].cov(pair["Volume"])
print(f"\nП.2г. Корреляция Close/Volume = {corr_cv:.4f}")
print(f"     Ковариация Close/Volume = {cov_cv:.4f}")

df_dropna = df.dropna(thresh=len(df.columns) - 1 - 1)
df_dropna = df[col_names].dropna(thresh=5)
print(f"\nПосле dropna(thresh=5): было {len(df)} строк, "
      f"стало {len(df_dropna)}")

df_filled = df[col_names].copy()

price_cols = ["Close", "Open", "High", "Low"]
row_mean = df_filled[price_cols].mean(axis=1)
for c in price_cols:
    df_filled[c] = df_filled[c].fillna(row_mean)

vol_mean = df_filled["Volume"].mean()
df_filled["Volume"] = df_filled["Volume"].fillna(vol_mean)

print("\nПропуски после fillna:")
print(df_filled.isna().sum())
print("\nПервые 12 строк после заполнения:")
print(df_filled.head(12))
```

На основании официальных статистических данных о курсе акции, которые были использованы при решении задания 3.2, а также данных о курсе акции за другой период (например, за прошлый год) осуществите:
а) соединение двух наборов данных в один, используя конкатенацию (concat());
б) соединение двух наборов данных в один, используя реляционный подход (merge()).



```python
print("П.3. Объединение наборов данных")

df_clean = df_filled.copy()

df_year = pd.DataFrame({
    "Date":  pd.to_datetime(["2025-01-10", "2025-02-10", "2025-03-10",
                             "2025-04-10", "2025-05-10"]),
    "Close": [6.20, 6.45, 6.90, 7.15, 7.30],
    "Open":  [6.10, 6.35, 6.80, 7.05, 7.20],
    "High":  [6.25, 6.50, 6.95, 7.20, 7.35],
    "Low":   [6.00, 6.25, 6.70, 6.95, 7.10],
    "Volume":[5_000_000, 4_800_000, 5_200_000, 4_500_000, 5_100_000],
})
print("\nНабор за прошлый год:", df_year.shape)

concat_year = pd.concat([df_year, df_clean], ignore_index=True)
print("concat (год + месяц):", concat_year.shape)

merge_year = pd.merge(df_year, df_clean, on="Date",
                      how="outer", suffixes=("_year", "_month"))
print("merge (outer по Date):", merge_year.shape)
```

На основании официальных статистических данных о курсе акций, которые были использованы при решении задания 3.2, а также данных о курсе акции за перекрывающий период (например, за текущий год) осуществите:
а) соединение двух наборов данных в один, используя конкатенацию, не допуская дублирования данных (concat());
б) соединение двух наборов данных в один, используя реляционный подход, не допуская дублирования данных (merge()).

```python
df_current = pd.DataFrame({
    "Date":  pd.to_datetime(["2026-09-09", "2026-09-10", "2026-09-11",
                             "2026-10-01", "2026-10-02", "2026-10-09"]),
    "Close": [10.150, 10.066, 10.300, 8.541, 8.432, 9.521],
    "Open":  [9.847, 10.233, 10.210, 8.856, 8.541, 9.185],
    "High":  [10.368, 10.340, 10.474, 8.856, 8.587, 9.590],
    "Low":   [9.800, 9.949, 9.981, 8.533, 8.342, 9.112],
    "Volume":[13_130_000, 9_070_000, 19_240_000, 14_090_000,
              10_140_000, 19_090_000],
})
print("\n3.2. Набор за текущий год (перекрытие):", df_current.shape)

concat_cur = pd.concat([df_clean, df_current], ignore_index=True)
concat_cur = concat_cur.drop_duplicates(subset=["Date"], keep="first")
print("concat без дублей:", concat_cur.shape)

merge_cur_inner = pd.merge(df_clean, df_current, on="Date",
                           how="inner", suffixes=("_m", "_c"))
print("merge (inner по Date, без дублей):", merge_cur_inner.shape)
print(merge_cur_inner)

merge_cur_outer = pd.merge(df_clean, df_current, on="Date", how="outer",
                           suffixes=("_m", "_c"))
merge_cur_outer = merge_cur_outer.drop_duplicates(subset=["Date"])
print("merge (outer + drop_duplicates):", merge_cur_outer.shape)
```

На основании официальных статистических данных о курсе акции, которые были использованы при решении задания 3.2, а также данных о курсе любой другой акции за аналогичный период осуществите:
а) соединение двух наборов данных в один, используя конкатенацию (concat());
б) соединение двух наборов данных в один, используя реляционный подход (merge()).

```python
df_sber = pd.DataFrame({
    "Date":   df_clean["Date"].values,
    "Close":  np.linspace(281, 301, len(df_clean)),
    "Open":   np.linspace(280, 300, len(df_clean)),
    "High":   np.linspace(282, 302, len(df_clean)),
    "Low":    np.linspace(278, 298, len(df_clean)),
    "Volume": np.linspace(8_000_000, 12_000_000, len(df_clean)),
})
print("\nНабор по другой акции за тот же период:",
      df_sber.shape)

df_mkb_t  = df_clean.copy();  df_mkb_t["Ticker"]  = "CBOM"
df_sber_t = df_sber.copy();   df_sber_t["Ticker"] = "SBER"

concat_two = pd.concat([df_mkb_t, df_sber_t], ignore_index=True)
print("concat (МКБ + Сбер):", concat_two.shape)

merge_two = pd.merge(df_mkb_t, df_sber_t, on="Date", how="inner",
                     suffixes=("_CBOM", "_SBER"))
print("merge (МКБ + Сбер по Date):", merge_two.shape)
print(merge_two.head(3))
```
