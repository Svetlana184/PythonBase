```python
import pandas as pd
import numpy as np
```

Создайте объект Series №1, содержащий названия пяти спортивных команд (тип данных – str, создание объекта осуществить с использованием конструктора, в качестве меток использовать ID команды, укажите имя объекта).

```python
s1 = pd.Series(
    ["Спартак", "Зенит", "ЦСКА", "Динамо", "Локомотив"],
    index=[101, 102, 103, 104, 105],
    name="Команды"
)
s1
```

Создайте объект Series №2, содержащий названия пяти городов (тип данных – str, создание объекта осуществить с использованием конструктора, в качестве меток использовать ID команды, укажите имя объекта).

```python
s2 = pd.Series(
    ["Москва", "Санкт-Петербург", "Москва", "Москва", "Москва"],
    index=[101, 102, 103, 104, 105],
    name="Города"
)
```

Создайте объект Series №3, содержащий количество побед пяти команд (тип данных – float, в качестве меток использовать ID команды).

```python
s3 = pd.Series(
    [12.0, 15.0, 10.0, 7.0, 9.0],
    index=[101, 102, 103, 104, 105],
    name="Победы"
)
```

Создайте объект Series №4, который содержит сведения о названии команды и городе.

```python
s4 = s1 + " — " + s2
s4.name = "Команда и город"
```

Осуществите следующие операции над объектом Series №3 (тип данных результирующих объектов – float): определите максимальное количество побед, определите минимальное количество побед, определите среднее количество побед.

```python
print("Максимум:", s3.max())
print("Минимум:", s3.min())
print("Среднее:", s3.mean())
```

Создайте объект DataFrame №1, содержащий следующую информацию о пяти спортивных командах: вид спорта, год основания, количество игроков (создание объекта осуществить с использованием конструктора, в качестве меток использовать ID команды).

```python
df1 = pd.DataFrame(
    {
        "Вид спорта":    ["Футбол", "Футбол", "Футбол", "Хоккей", "Футбол"],
        "Год основания": [1922, 1925, 1911, 1946, 1922],
        "Игроков":       [25, 24, 23, 22, 26],
    },
    index=[101, 102, 103, 104, 105]
)
print(df1)
```

Создайте объект DataFrame №2 на основе объединения нескольких объектов Series: Series №1, Series №2, Series №3.

```python
df2 = pd.DataFrame({
    "Команда": s1,
    "Город":   s2,
    "Победы":  s3,
})
print(df2)
```

Создайте объект DataFrame №3 на основе объединения нескольких объектов DataFrame: DataFrame №1, DataFrame №2.

```python
df3 = pd.concat([df1, df2], axis=1)
print(df3)
```

Осуществите вывод на экран для объекта DataFrame №3 списка меток строк, списка меток столбцов, массива данных (по отдельности).

```python
print("Метки строк:   ", df3.index.tolist())
print("Метки столбцов:", df3.columns.tolist())
print("Массив данных:\n", df3.values)
```
