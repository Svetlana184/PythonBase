Создание данных типа "класс". Поля x, y, вычисление куба наибольшего из чисел

```python
class Cube:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def get_info(self):
        print("x:", self.x)
        print("y:", self.y)
    def get_cube(self):
        if self.x > self.y:
            print("Куб числа x:", self.x*self.x*self.x)
        else:
            print("Куб числа y:", self.y*self.y*self.y)

cube = Cube(1, 2)
cube.get_cube()
```

Наследование. Произведение числа z на минимальное из чисел x и y (полей класса родителя)

```python
class ZCube (Cube):
    def __init__(self, x, y, z):
        super().__init__(x, y)
        self.z = z
    def composition(self):
        if self.x < self.y:
            print("Произведение числа z на x:", self.x * self.z)
        if self.x > self.y:
            print("Произведение числа z на y:", self.y * self.z)

cube1 = ZCube(3,4,5)
cube1.composition()
```

Полиморфизм.
Первый класс: высшее учебное заведение (название заведения, количество первокурсников, количество выпускников,
q = выпускники / первокурсники). Дочерний класс: дополнительное поле "процент выпускников, работающих по специальности",
дополнительная функция q * дополнительное поле

```python
class Uni:
    def __init__(self, uni_name, graduates, first_years):
        self.uni_name = uni_name
        self.graduates = graduates
        self.first_years = first_years

    def get_stats(self):
        return self.graduates / self.first_years

    def get_info(self):
        print("=============== ИНФОРМАЦИЯ О ВУЗЕ ==============")
        print("название:", self.uni_name)
        print("количество выпускников:", self.graduates)
        print("количество первокурсников:", self.first_years)
        print("выпускники / первокурсники:", self.get_stats())

class SuperUni(Uni):
    def __init__(self, uni_name, graduates, first_years, percent_working):
        super().__init__(uni_name, graduates, first_years)
        self.percent_working = percent_working
    def get_info(self):
        super().get_info()
        print("процент выпускников, работающих по специальности:", self.percent_working)
    def get_super_stats(self):
        return self.get_stats() * self.percent_working


uni1 = Uni('kbk', 100, 50)
uni1.get_info()
uni2 = SuperUni('kgty', 100, 50, 44)
uni2.get_info()
```
