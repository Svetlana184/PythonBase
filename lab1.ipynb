Задача 1. Все отрицательные элементы массива заменить на их абсолютную величину

```python
from random import randint
mas = tuple(randint(-100, 100) for i in range (randint(10, 21)))

print("Изначальный кортеж: ")

for i in mas:
    print(i, end=" ")

print("\nКортеж после обработки: ")

mas_positive = tuple(i if i > 0 else -i for i in mas)
for i in mas_positive:
     print(i, end=" ")
```

Задача 2. Определить, является ли квадратный массив симметричным относительно своей главной диагонали

```python
n = 3
matrix = tuple(tuple(randint(1, 2) for j in range (n)) for i in range (n))

for i in matrix:
    print(*i)

symmetry = "массив симметричен"

for i in range(len(matrix)):
    for j in range(len(matrix)):
        if matrix[i][j] != matrix[j][i]:
            symmetry = "массив не симметричен"
            break

print(symmetry)
```

Задача 3. Дан список чисел. Выведите значение наибольшего элемента в списке, а затем индекс этого элемента в списке.
Если наибольших элементов несколько, выведите индекс первого из них

```python
mas3 = [randint(10, 100) for i in range (randint(10, 21))]
max_number = 0
max_index = 0

print("Массив: ")

for i in mas3:
    print(i, end=" ")

for i in range(len(mas3)):
    if mas3[i] > max_number:
        max_number = mas3[i]
        max_index = i

print("\nЗначение наибольшего элемента в массиве: ", max_number, ", индекс элемента: ", max_index)
```

 Задача 4. Дан текст: в первой строке записано число строк, далее идут сами строки. Определите, сколько различных слов содержится в этом тексте.
 Словом считается последовательность непробельных символов, идущих подряд, слова разделены одним иои большим количеством пробелов или символами конца строки

```python
text = ("5\n"
        "ААА ааа а.\n"
        "B bbb b.\n"
        "yy y?\n"
        "t tttt.\n"
        "i i i.\n")

lines = text.strip().split("\n")
lines = lines[1:]
joined_text = ' '.join(lines).replace(".", " ").replace(",", " ").replace("!", " ").replace("?", " ")
words = joined_text.split()
text_set = set(words)

print(text_set)
```

Задача 5. Напишите программу, которая подсчитывает количество единиц товаров, приобретенных покупателями онлайн-магазина.
На вход программе подается число n - количество записей о покупках, а затем n строк вида "Покупатель товар количество".
Для каждого покупателя программа должна выводить список покупок

```python
n = int(input("Введите число записей: "))
mas_goods = dict()

for i in range(n):
    good = input("Введите запись").strip().split(" ")
    mas_goods[good[0]] = good[1] + " " + good[2]

byer = input("введите имя покупателя")

print(mas_goods[byer])
```
